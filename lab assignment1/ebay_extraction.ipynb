```python
#NAME: KAMINY VARMA || ID:IS01083876
#NAME: LEENA DEWI THEVAN || ID:IS01083877
#NAME:ZALIN ZALIFAH BT ZULKIFLI || ID:IS01083447
#PAGE1
from selenium import webdriver
from selenium.webdriver.chrome.service import Service
from webdriver_manager.chrome import ChromeDriverManager
from selenium.webdriver.common.by import By
import time

# 1. Setup Driver
options = webdriver.ChromeOptions()
# added a mobile user agent because URL is 'mweb'
options.add_argument("--user-agent=Mozilla/5.0 (iPhone; CPU iPhone OS 14_6 like Mac OS X) AppleWebKit/605.1.15 (KHTML, like Gecko) Version/14.0.3 Mobile/15E148 Safari/604.1")
driver = webdriver.Chrome(service=Service(ChromeDriverManager().install()), options=options)

target_url = "https://www.ebay.com/fdbk/mweb_profile?fdbkType=FeedbackReceivedAsSeller&item_id=336400385965&username=integritytech&filter=feedback_page%3ARECEIVED_AS_SELLER&q=336400385965&sort=RELEVANCEV2"

try:
    print(f"Opening: {target_url}")
    driver.get(target_url)
    
    # Wait for the mobile feedback cards to load
    print("Waiting for eBay feedback to load...")
    time.sleep(8) 

    # 2. TARGET THE FEEDBACK CARDS
    # eBay mobile feedback often uses 'fdbk-container' or specific table rows
    feedback_cards = driver.find_elements(By.CSS_SELECTOR, ".fdbk-container__details")
    
    if not feedback_cards:
        # Fallback for different mobile layouts
        feedback_cards = driver.find_elements(By.CLASS_NAME, "card")

    if len(feedback_cards) > 0:
        print(f"SUCCESS: Found {len(feedback_cards)} feedback entries.")
        # Display raw sample to check
        print("\n--- RAW DATA SAMPLE ---")
        print(feedback_cards[0].text)
    else:
        print("Still 0 entries found. eBay might be blocking the mobile view or requiring a login.")

except Exception as e:
    print(f"Error: {e}")



# Name: Kaminy Varma | ID: IS01083876
import csv
from selenium.webdriver.common.by import By

def organize_and_save(feedback_cards):
    """
    Purpose: Separates raw text into Name, Date, and Content, 
    then saves to a CSV file.
    """
    organized_list = []

    print(f"Organizing {len(feedback_cards)} entries...")

    for card in feedback_cards:
        try:
            # Extract raw text from the card
            raw_text = card.text
            lines = raw_text.split('\n')
            
            # Purpose: Map the lines to the correct fields
            # Based on your sample:
            # Line 0: Reviewer Name (e.g., r***e)
            # Line 2: Date (e.g., Past month)
            # Line 4: Content (The actual review text)
            
            reviewer_name = lines[0] if len(lines) > 0 else "N/A"
            review_date = lines[2] if len(lines) > 2 else "N/A"
            # We join remaining lines to ensure we get the full comment
            review_content = " ".join(lines[4:]) if len(lines) > 4 else "N/A"

            organized_list.append({
                "Reviewer Name": reviewer_name,
                "Review Date": review_date,
                "Review Content": review_content
            })
        except Exception as e:
            continue

    # Step 3: Write to CSV
    filename = "ebay_feedback.csv"
    keys = ["Reviewer Name", "Review Date", "Review Content"]
    
    with open(filename, 'w', newline='', encoding='utf-8') as f:
        writer = csv.DictWriter(f, fieldnames=keys)
        writer.writeheader()
        writer.writerows(organized_list)
    
    print(f"SUCCESS! {len(organized_list)} reviews saved to {filename}.")

# Call this function using the 'feedback_cards' variable from previous code
organize_and_save(feedback_cards)
    
```
